Create a new column in the df refer the miles column

```python
#Create New Column based on the miles if miles > 10 then 'Long Drive' else 'Short Drive'
def compare_miles(miles):
    if miles > 10:
        return 'Long Drive'
    elif miles >5 and miles <= 10:
        return 'Average Drive'
    else:
        return 'Short Drive'
    
df_dropped['DRIVE_TYPE'] = df_dropped['MILES*'].apply(compare_miles)
df_dropped.loc[0:9, ['MILES*','DRIVE_TYPE']]
```

#Data Cleaning

## Handle Missing Value

1. Case 1 - If any col has % of missing values more than a threashold, drop that col
2. Case 2 - Drop all missing value using dropna, drop from sel col  -- BE CAREFUL SINCE DROPNA drops the row even if one column is having Empty value
3. Case 3 - Fill the missing value
            - Numerical col -- nmber, mean, median, constant
            -categorical data -- mode, NA, Constant, safe Choice is always NA if we dont real data or purpose

```python
#Creat copy of DataFrame
df_copy = df_dropped.copy()
df_copy.shape

#drop all the missing values rows
df_no_missing = df_copy.dropna() #drops all rows with any missing values and assigns to new DataFrame
df_copy.dropna(inplace=True)  #drops all rows with any missing values in the same DataFrame, no need to assign to new DataFrame, changes are permanent
df_copy.shape
```

```python
#Drop missing value from specific columns, Use the Subset to specify the columns to check for missing values
df_copy_columns=df_dropped.copy()
df_copy_columns.dropna(subset=['PURPOSE*',"MILES*"], inplace=True) 
df_copy_columns.shape
```

```python
#Fill the Missing Values
df_dropped["PURPOSE*"].fillna("NA", inplace=True)
df_dropped.info()
```

#Sort the Data

```python
#Sorting the DataFrame by MILES* in ascending order
df_sorted = df_dropped.sort_values(by='MILES*')
df_sorted

#sorting the DataFrame by MILES* in descending order
df_sorted_desc = df_dropped.sort_values(by='MILES*', ascending=False)
df_sorted_desc


#Using more than 2 columns for sorting
df_sorted_multi = df_dropped.sort_values(by=['START*','MILES*'], ascending=[True, False])
df_sorted_multi
```

Group By

```python
#Grouping the DataFrame, Group by START* column and count number of entries in MILES* column for each group
df_grouped = df_dropped.groupby('START*')["MILES*"].count()
df_grouped
```

```python
#Print sum of miles for each START* location
df_grouped_sum = df_dropped.groupby('START*')["MILES*"].sum()
df_grouped_sum
```

```python
#Print Avg of miles for each START* location
df_grouped_avg = df_dropped.groupby('START*')["MILES*"].mean()
df_grouped_avg
```

```python
#Print the avg miles for each START* location in sorted order
df_grouped_avg_sorted = df_dropped.groupby('START*')["MILES*"].mean().sort_values(ascending=True)  
df_grouped_avg_sorted
```

```python
#Print the avg miles for each START* location miles >1
df_grouped_avg_filtered = df_dropped.groupby('START*')["MILES*"].mean()
df_grouped_avg_filtered = df_grouped_avg_filtered[df_grouped_avg_filtered > 1]
df_grouped_avg_filtered

 
```

```python
#If onky sum is used then all numeric columns are summed up
df_grouped_sum_all = df_dropped.groupby('START*').sum(numeric_only=True) 
df_grouped_sum_all
```

```python
#Show mean, sum, count, min, max for MILES* column grouped by START* location
df_grouped_stats = df_dropped.groupby('START*')["MILES*"].agg(['mean', 'sum', 'count', 'min', 'max'])
df_grouped_stats
```

```python
#Sort by mean
df_grouped_stats = df_dropped.groupby('START*')["MILES*"].agg(['mean', 'sum', 'count', 'min', 'max']).sort_values(by='mean')
df_grouped_stats
```

```python
#Print top 5 start stats from where the many drives started, use nlargest function
df_grouped_stats = df_dropped.groupby('START*')
df_grouped_stats.size().nlargest(2 )
```

```python
#Create group using 2 columns START* and CATEGORY*
df_grouped_two = df_dropped.groupby(['START*', 'CATEGORY*','PURPOSE*'])
df_grouped_two.size()
```
