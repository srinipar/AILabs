Create a new column in the df refer the miles column

```python
#Create New Column based on the miles if miles > 10 then 'Long Drive' else 'Short Drive'
def compare_miles(miles):
    if miles > 10:
        return 'Long Drive'
    elif miles >5 and miles <= 10:
        return 'Average Drive'
    else:
        return 'Short Drive'
    
df_dropped['DRIVE_TYPE'] = df_dropped['MILES*'].apply(compare_miles)
df_dropped.loc[0:9, ['MILES*','DRIVE_TYPE']]
```

#Data Cleaning

## Handle Missing Value

1. Case 1 - If any col has % of missing values more than a threashold, drop that col
2. Case 2 - Drop all missing value using dropna, drop from sel col  -- BE CAREFUL SINCE DROPNA drops the row even if one column is having Empty value
3. Case 3 - Fill the missing value
            - Numerical col -- nmber, mean, median, constant
            -categorical data -- mode, NA, Constant, safe Choice is always NA if we dont real data or purpose

```python
#Creat copy of DataFrame
df_copy = df_dropped.copy()
df_copy.shape

#drop all the missing values rows
df_no_missing = df_copy.dropna() #drops all rows with any missing values and assigns to new DataFrame
df_copy.dropna(inplace=True)  #drops all rows with any missing values in the same DataFrame, no need to assign to new DataFrame, changes are permanent
df_copy.shape
```

```python
#Drop missing value from specific columns, Use the Subset to specify the columns to check for missing values
df_copy_columns=df_dropped.copy()
df_copy_columns.dropna(subset=['PURPOSE*',"MILES*"], inplace=True) 
df_copy_columns.shape
```

```python
#Fill the Missing Values
df_dropped["PURPOSE*"].fillna("NA", inplace=True)
df_dropped.info()
```
