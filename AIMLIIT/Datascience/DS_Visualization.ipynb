```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#line CHarts, Single Line
x = np.array([100,200,300,400,500])
y = np.random.randint(1,100,size=5)
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
plt.title("Line Chart Example")

plt.plot(x,y, marker='*', color='b', linestyle='--', linewidth=2, markersize=38)
plt.grid(True)
plt.xticks(x)
plt.show()
```

```python
#line Charts, Multiple Lines - 3 lines
x = np.array([1,2,3,4,5])
y1 = np.random.randint(1,100,size=5)
y2 = np.random.randint(1,100,size=5)
y3 = np.random.randint(1,100,size=5)
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")

plt.title("Multiple Line Chart Example")
plt.plot(x,y1, marker='o', color='r', linestyle='--', linewidth=2, markersize=10, label='Jan')
plt.plot(x,y2, marker='*', color='b', linestyle='-.', linewidth=2, markersize=10, label='Feb')
plt.plot(x,y3, marker='s', color='g', linestyle=':', linewidth=2, markersize=10, label='Mar')
plt.grid(True)
plt.xticks(x) 
plt.legend(loc="center right")
plt.show()
```
