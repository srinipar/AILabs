```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#line CHarts, Single Line
x = np.array([100,200,300,400,500])
y = np.random.randint(1,100,size=5)
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
plt.title("Line Chart Example")

plt.plot(x,y, marker='*', color='b', linestyle='--', linewidth=2, markersize=38)
plt.grid(True)
plt.xticks(x)
plt.show()
```

```python
#line Charts, Multiple Lines - 3 lines
x = np.array([1,2,3,4,5])
y1 = np.random.randint(1,100,size=5)
y2 = np.random.randint(1,100,size=5)
y3 = np.random.randint(1,100,size=5)
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")

plt.title("Multiple Line Chart Example")
plt.plot(x,y1, marker='o', color='r', linestyle='--', linewidth=2, markersize=10, label='Jan')
plt.plot(x,y2, marker='*', color='b', linestyle='-.', linewidth=2, markersize=10, label='Feb')
plt.plot(x,y3, marker='s', color='g', linestyle=':', linewidth=2, markersize=10, label='Mar')
plt.grid(True)
plt.xticks(x) 
plt.legend(loc="center right")
plt.show()
```

Univariate Analysis Chart

```python
#histogram
data = np.random.normal(134, 20, 450) # mean=134, std=20, n=450
plt.hist(data, bins=30, color='c', edgecolor='black')
plt.title("Histogram Example")
plt.show()
```

```python
#Box Plot, Shows the 5 number summary of the data (Lower, Q1,Q2,Q3,upper)
data=np.random.normal(134,20,450)# mean=134, std=20, n=450
plt.boxplot(data)
plt.title("Box Plot Example")
plt.show()
```

Bivariate Analysis

```python
#scatter Plot, num vs num, area vs house price etc
Age = np.array([25,30,35,40,45,50,55,60,65,70,75,80])
Salary = np.array([35000,40000,45000,50000,60000,65000,70000,80000,85000,90000,95000,100000])           
plt.scatter(Age, Salary, color='m', marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

#Categorical Data vs Numeric Data

```python
#Bar Chart, categorical vs numerical
fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']
values = [10, 20, 15, 25, 30]
plt.bar(fruits, values)
plt.plot(fruits, values, marker='o', color='r', linestyle='--', linewidth=2, markersize=8)
plt.title("Bar Chart Example")
plt.xlabel("Fruits")
plt.ylabel("Values")
plt.grid(True)
plt.show()
```

```python
#sale in 3 regions
products = ['Product A', 'Product B', 'Product C', 'Product D']
region1_sales = [100, 150, 200, 250]
region2_sales = [120, 180, 220, 280]
region3_sales = [90, 140, 190, 240]

plt.bar(products,region1_sales, label='Region 1')
plt.bar(products,region2_sales, bottom=region1_sales, label='Region 2')
plt.bar(products,region3_sales, bottom=np.array(region1_sales)+np.array(region2_sales), label='Region 3')

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales in Three Regions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Stack Plot
days=[1,2,3,4,5]
social_media_traffic=[100,120,150,170,200]
add_traffic=[80,90,100,110,130]
reference_traffic=[50,60,70,80,90]

plt.stackplot(days, social_media_traffic, add_traffic, reference_traffic, labels=['Social Media','Ads','Reference'], colors=['m','c','y'])
plt.xlabel("Days")
plt.ylabel("Traffic")
plt.title("Website Traffic Sources Over Days")
plt.legend(loc='upper left')
plt.show()
```

```python
#pie Chart
labels=['Python','Java','C++','JavaScript']
sizes=[45,30,15,10]
explode = (0.1, 0, 0, 0)  # explode the 1st slice (Python)
plt.pie(sizes, labels=labels, autopct='%1.1f%%', explode=explode)
plt.title("Programming Language Distribution")
plt.show()
```

Put more Charts in One Output

```python
#Put more Charts in One Figure using subplot
x = np.array([1,2,3,4,5])
y1 = np.random.randint(1,100,size=5)
y2 = np.random.randint(1,100,size=5)
y3 = np.random.randint(1,100,size=5)

plt.figure(figsize=(12, 4))

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title("Line Plot 1")

plt.subplot(2, 2, 2)
plt.bar(x, y2)
plt.title("Bar Plot")

plt.subplot(2, 2, 3)
plt.scatter(x, y3)
plt.title("Scatter Plot")

plt.subplot(2, 2, 4)
plt.scatter(x, y3)
plt.title("Scatter Plot")

plt.tight_layout()
plt.savefig("multiple_charts.png")
plt.show()
```
