```python
#

def detect_outliers(data,column):
    Q1=data[column].quantile(0.25)
    Q3=data[column].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-(1.5*IQR)
    upper_bound=Q3+(1.5*IQR)

    outliers=data[(data[column]<lower_bound)| data[column]>upper_bound]
    print(outliers)
    print(f"Q1 {Q1} Q2{Q3}, IQR {IQR}")
    print(f"Lower_Bound {lower_bound}, Upper_Bound {upper_bound}")
    print("No of Outliers detected in col", len(outliers))
    return outliers

outliers_rows=detect_outliers(tips, 'total_bill')
```

```python
#Remove Outliers from the Data
#Any data greater than the lower Bound and less than the upper bound is called safe data, opposite is outliers
def remove_outlier_iqr(data,column):
    Q1=data[column].quantile(0.25)
    Q3=data[column].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-(1.5*IQR)
    upper_bound=Q3+(1.5*IQR)

    new_data=[(data[column])>=lower_bound & (data[column]<=upper_bound)]
    return new_data

tips_cleaned=remove_outlier_iqr(tips, 'total_bill')
tips_cleaned.shape
```
