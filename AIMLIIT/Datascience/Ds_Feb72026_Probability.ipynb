##Probability

```python
import pandas as pd
import numpy as np
print("Imported")
```

```python
data=pd.DataFrame({"Clicked":[1,1,0,0,0,1,1,0,0,0]}) #10 Users

p_hat=data.mean()

p_hat
```

```python
#Simulate this data for 100 Customers

from scipy.stats import bernoulli

simulated = bernoulli.rvs(p_hat, size=10000)
print(f"simulated Clicked Rate :", simulated.mean() )
```

```python
#Plot
import matplotlib.pyplot as plt

#visualize
plt.hist(simulated, bins=np.arange(-0.5, 2,1), density=True, color='blue', linewidth=1.5)
plt.xticks([0,1], ['No Click', 'Click'])
plt.title('Simulated Email Click Behaviour')
plt.xlabel('User Action')
plt.ylabel('Probability')
plt.show()
```
