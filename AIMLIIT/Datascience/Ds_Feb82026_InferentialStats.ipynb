T-Test

One Sample T-test- The One sample T test is used to compare the mean of a sample with a known population meann., This test employs the standard deviation of the sample instead of the population standard deviation due to the unknown piopulation variance, the formula for a one-sample t-test is as follows

```python
ages=pd.read_csv(r"Datasets/Feb 8 2026_Hypothesis/Ages.csv")

from scipy.stats import ttest_1samp

age_col=ages["ages"]

#before Test
print("Mean age of emp" , age_col.mean())
print("std of emp age ", age_col.std() )

```

```python
#Perform Test

t_stat, p_value=ttest_1samp(age_col,30)
print(f'p-value:{p_value}')
print(f'test_stat:{t_stat}')
```

```python
alpha=0.05
if p_value<alpha:
    print('Reject the null hypothesis mean is not 30')
else:
    print("Fail to reject the null hypothesis(accpet HO) mean is 30")
```
