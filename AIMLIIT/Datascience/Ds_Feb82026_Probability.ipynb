#Uniform Distribution
    equal prob == all values will have equal chance
    Bus Arrival time --0 to 10, all times are equally likely, uniform distribution
    

```python
#Uniform Distrubution
from scipy.stats import uniform

#Generate Uniformly Ditributed Data

a,b=0,10 #Define Start and end points of the uniform distribution

data_uniform=uniform.rvs(a,b-1, size=10000)

#Plot histogram
plt.hist(data_uniform, bins=30, density=True, alpha=0.5, color='blue')

#calcualte the PDF of the uniform distribution

uniform_xs=np.linspace(a,b,301)
uniform_ys=uniform.pdf(uniform_xs,a,b-1)

#plot PDF
plt.plot(uniform_xs, uniform_ys, color='red')

#add labels
plt.xlabel('Uniform Distribution')
plt.ylabel('Density')

#display the plot
plt.show()
```
