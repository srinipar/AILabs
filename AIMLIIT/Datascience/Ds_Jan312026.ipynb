```python
#%pip install plotly
#%pip install --upgrade nbformat
import pandas as pd
import plotly.express as px
import numpy as np
```

```python
df=px.data.iris()
 
#histogram
fig=px.histogram(df,x='sepal_width',nbins=20,title='Sepal Width Distribution')
fig.show()
```

```python
#Create a scatter plot
fig=px.scatter(df,x='sepal_length',y='sepal_width',color='species',title='Sepal Length vs Sepal Width')
fig.show()
```

```python
#Create a box plot
fig=px.box(df,x='species',y='petal_length',title='Petal Length by Species')
fig.show()
```

```python
#create a scatter matrix
fig=px.scatter_matrix(df,dimensions=['sepal_length','sepal_width','petal_length','petal_width'],color='species',title='Scatter Matrix of Iris Dataset')
fig.show()
```

```python
#Heatmap
corr=df.corr(numeric_only=True)
fig=px.imshow(corr,labels=dict(x="Features", y="Features", color="Correlation"), x=corr.columns, y=corr.columns)
fig.show()  
```

```python
#density heatmap
corr=df.corr(numeric_only=True)
fig=px.density_heatmap(df,x='sepal_length',y='sepal_width',nbinsx=20,nbinsy=20,title='Density Heatmap of Sepal Length and Width')
fig.show()
```

```python
import seaborn as sns
tips=sns.load_dataset('tips')

tips['day'].value_counts()
tips['day'].value_counts().plot(kind='bar', backend='plotly',title='Number of Tips by Day') #backend='plotly' to use plotly for plotting default is matplotlib
```
