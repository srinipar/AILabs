```python
import pandas as pd
#Create Series Data

Series1 = pd.Series([10,20,30,40,50])
print(Series1) #Since all the data are int type it will consider as int64 datatype
print(Series1.dtype) #int64
```

```python

Series2 = pd.Series([10,20,30,40,50,1.0])
print(Series2) #Since all the data are int, float  type it will consider as float datatype
print(Series2.dtype) #float64
```

```python
Series3 = pd.Series([10,20,30,40,50,1.0, "Srini"])
print(Series3) #Since all the data are int, Object  type it will consider as float datatype
print(Series3.dtype) #Object
print(type(Series3[0])) #each value type is preserved int or based on the value it will consider the type
```

```python
#Set the Index for the Series, it can be string or numbers
ser=pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
print(ser)
```

```python
#Create a series using numpy array
import numpy as np
arr = np.array([10,20,30,40,50])
print(arr) #Index will be created automatically from 0 to n-1
series_from_array = pd.Series(arr, index=['a','b','c','d',2]) #Convert numpy array to pandas series
print(series_from_array)
```

```python
#Create a Series using Dictionary, keys will be considered as index, values as data
data = {'a':10, 'b':20, 'c':30, 'f':40}
series_from_dict = pd.Series(data)  
print(series_from_dict) #Even Index can be modified while creating Series from Dictionary
series_from_dict2 = pd.Series(data, index=['a','b','c','d','e'])  
print(series_from_dict2) #If index not found in dictionary it will consider as NaN
```

```python
#Functions on Series
ser1=pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
ser2=pd.Series([10,20,30,40,50], index=['a','b','c','d','e'])

print(f"Addition of two Series: \n{ser1 + ser2}")  #Addition of two Series based on index
print(f"Subtraction of two Series: \n{ser2 - ser1}")    #Subtraction of two Series based on index
print (ser1.add(ser2))  #Addition using method
print (ser2.sub(ser1))  #Subtraction using method

#WHen Index is not same
ser3=pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
ser4=pd.Series([10,20,30,40,50], index=['a','f','b','y','z'])
print(f"Addition of two Series with different index: \n{ser3 + ser4}")  #Addition of two Series based on index, missing index will be NaN
print (ser4.add(ser3))  #Addition using method, missing index will be
```
