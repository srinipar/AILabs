```python
import pandas as pd
import numpy as np
```

```python
#Joins/merges between two DataFrames

first_df=pd.DataFrame({'ID':[1,2,3,4],
'Name':['Alice','Bob','Charlie','David'],
'Age':[25,30,35,40]})

second_df=pd.DataFrame({'ID':[1,2,3,4],
'Name':['A','B','C','D'],
'Age':[10,20,30,50]})

merged_df=pd.merge(first_df,second_df,on='ID',how='inner',suffixes=('_First','_Second'))
print(merged_df)
```

Concat

```python
#Concatination of DataFrames , Ignore index will not consider the previous index values so it becomes as continuous
#if we dont need the index to change use the key to define the source of data
#axis =0 means row wise concatination , axis=1 means column wise concatination
concat_df=pd.concat([first_df,second_df], keys=['First_DF','Second_DF'])
(concat_df)

concat_df.loc['First_DF']
```
