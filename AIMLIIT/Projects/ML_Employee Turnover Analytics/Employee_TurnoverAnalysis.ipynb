```python
#imported libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
#load the Dataset of HR_comma_sep.csv
data = pd.read_csv("Datasets/HR_comma_sep.csv")
data.head()
```

```python
#Check the missing values in the dataset
print(data.isnull().sum()) #no null values in the dataset
#Check the shape of the dataset
print(data.shape) #14999 rows and 10 columns in the dataset
print(data.info()) #check the data types of the columns in the dataset
```

```python
#2.1.	Draw a heatmap of the correlation matrix between all numerical features or columns in the data.
num_vars = data.select_dtypes(include=['int64', 'float64']).columns
num_vars

#create the correlation matrix
corr_matrix = data[num_vars].corr()

#plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
#2.2.	Draw the distribution plot of:
#■	Employee Satisfaction (use column satisfaction_level)
#■	Employee Evaluation (use column last_evaluation)
#■	Employee Average Monthly Hours (use column average_montly_hours)

#Distribution plot of Employee Satisfaction
plt.figure(figsize=(8, 6))
sns.histplot(data['satisfaction_level'], kde=True, bins=30, color='blue')
plt.title('Distribution of Employee Satisfaction')
plt.xlabel('Satisfaction Level')
plt.ylabel('Frequency')
plt.show()
#Distribution plot of Employee Evaluation
plt.figure(figsize=(8, 6))
sns.histplot(data['last_evaluation'], kde=True, bins=30, color='green')
plt.title('Distribution of Employee Evaluation')
plt.xlabel('Last Evaluation')
plt.ylabel('Frequency')
plt.show()

#Distribution plot of Employee Average Monthly Hours
plt.figure(figsize=(8, 6))
sns.histplot(data['average_montly_hours'], kde=True, bins=30, color='orange')
plt.title('Distribution of Employee Average Monthly Hours')
plt.xlabel('Average Monthly Hours')
plt.ylabel('Frequency')
plt.show()

```
