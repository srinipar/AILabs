```python
#imported libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
#load the Dataset of HR_comma_sep.csv
data = pd.read_csv("Datasets/HR_comma_sep.csv")
data.head()
```

```python
#Check the missing values in the dataset
print(data.isnull().sum()) #no null values in the dataset
#Check the shape of the dataset
print(data.shape) #14999 rows and 10 columns in the dataset
print(data.info()) #check the data types of the columns in the dataset
```

```python
#2.1.	Draw a heatmap of the correlation matrix between all numerical features or columns in the data.
num_vars = data.select_dtypes(include=['int64', 'float64']).columns
num_vars

#create the correlation matrix
corr_matrix = data[num_vars].corr()

#plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
#2.2.	Draw the distribution plot of:
#■	Employee Satisfaction (use column satisfaction_level)
#■	Employee Evaluation (use column last_evaluation)
#■	Employee Average Monthly Hours (use column average_montly_hours)

#lets use Subplots to draw the distribution plots of the three features in a single figure
plt.figure(figsize=(18, 5))


#Distribution plot of Employee Satisfaction
plt.subplot(1, 3, 1)
sns.histplot(data['satisfaction_level'], kde=True, bins=30, color='blue')
plt.title('Distribution of Employee Satisfaction')
plt.xlabel('Satisfaction Level')
plt.ylabel('Frequency')
#Distribution plot of Employee Evaluation
plt.subplot(1, 3, 2)
sns.histplot(data['last_evaluation'], kde=True, bins=30, color='green')
plt.title('Distribution of Employee Evaluation')    
plt.xlabel('Last Evaluation')
plt.ylabel('Frequency')
#Distribution plot of Employee Average Monthly Hours
plt.subplot(1, 3, 3)
sns.histplot(data['average_montly_hours'], kde=True, bins=30, color='orange')
plt.title('Distribution of Employee Average Monthly Hours')
plt.xlabel('Average Monthly Hours')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

```

3.	Perform clustering of employees who left based on their satisfaction and evaluation.

    3.1.	Choose columns satisfaction_level, last_evaluation, and left.

    3.2.	Do K-means clustering of employees who left the company into 3 clusters?

    3.3.	Based on the satisfaction and evaluation factors, give your thoughts on the employee clusters.


```python
#people left the company.
left_count = data['left'].value_counts()
print(left_count)

#3571 employees left the company and 11428 employees stayed in the company.

#Clustering based on statisfaction level and last evaluation those who left
left_employees = data[data['left'] == 1]
clustering_data = left_employees[['satisfaction_level', 'last_evaluation']]
print("Shape of the data for clustering:", clustering_data.shape    )
print(clustering_data.head())
```

```python
#Perform KMeans clustering on the data of those who left the company based on satisfaction level and last evaluation
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(clustering_data)
```

```python
kmeans_labels = kmeans.labels_
print("Cluster labels assigned to each employee who left:", kmeans_labels)
#append the cluster labels to the original data of those who left
clustering_data['Cluster'] = kmeans_labels
print(clustering_data.head())
```

```python
#Visualize the clusters using a scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='satisfaction_level', y='last_evaluation', hue='Cluster', data=clustering_data, 
                palette=['red', 'green', 'blue'])
plt.title('KMeans Clusters of Employees Who Left')
plt.xlabel('Satisfaction Level')
plt.ylabel('Last Evaluation')
plt.show()
```

```python
#observations from the KMeans clustering:
#1. Blue Cluster: This cluster represents employees with low satisfaction levels and low last evaluation scores. These employees may have left the company due to dissatisfaction with their job or poor performance evaluations.
#2. Green Cluster: This cluster represents employees with moderate satisfaction levels and moderate last evaluation scores. These employees may have left the company due to a combination of factors.
#3. Red Cluster: This cluster represents employees with high satisfaction levels and high last evaluation scores. These employees are less likely to leave the company.
```
