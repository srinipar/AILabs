```python
#Calculator Function to add, subtract, multiply, divide
def calculator(num1: float, num2: float, operation: str = '+') -> float:
    '''Performs basic arithmetic operations'''
    if operation == '+':
        return num1 + num2
    elif operation == '_':
        return num1 - num2
    elif operation == '*':
        return num1 * num2
    elif operation == '/':
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero"
    else:
        return "Error: Invalid operation"
#Testing the calculator function
exit_flag = False# Loop until user decides to exit
while not exit_flag:
    input1 =  float(input("Enter first number: "))
    input2 = float(input("Enter second number: "))
    op = input("Enter operation (+, -, *, /): ")
    result = calculator(input1, input2, op)
    print(f"Result: {result}")
    cont = input("Do you want to perform another calculation? (yes/no): ").strip().lower()
    if cont != 'yes':
        exit_flag = True
```

```python
#Functions with *args, arguments are saved as a tuple, you cannot have normal arguments after *args
def func_with_args(*args):
    '''Function demonstrating *args '''
    print("Positional arguments (args):", args)
    for arg in args:
        print("Argument:", arg)
    
func_with_args(1, 2, 3, 'Srini', [4, 5, 6])
```

```python
#functions with **kwargs, arguments are saved as a dictionary
def func_with_kwargs(**kwargs):
    '''Function demonstrating **kwargs '''
    print("Keyword arguments (kwargs):", kwargs)
    for key, value in kwargs.items():
        print(f"Key: {key}, Value: {value}")

func_with_kwargs(name="Srini", age=26, course=["ML", "DL", "AI"])
```

```python
#functions with parametes, normal, *args and **kwargs
def func_with_all_params(normal_param, *args, **kwargs):
    '''Function demonstrating normal parameter, *args and **kwargs '''
    print("Normal parameter:", normal_param)
    print("Positional arguments (args):", args)
    print("Keyword arguments (kwargs):", kwargs)

func_with_all_params("NormalValue", 1, 2, 3, name="Srini", age=26)
```

```python
#Lambda Functions, This will work only for simple operations 
#Lambda function to add two numbers
def add (x, y): return x + y
result = add(5, 3)
print("Sum:", result)

add = lambda x=0, y=0: x + y
result = add(10, 20)    
print("Sum:", result)
```

```python
#MAP, call the function without loop and should be assigned to variable container like list, set, tuple
sq=lambda x: x * 2
map_result = map(sq, [1, 2, 3, 4, 5]) #Doubling each element in the list
print("Map Result:", list(map_result))
```

```python
#Filter, call the function without loop and should be assigned to variable container like list, set, tuple
is_even=lambda x: x % 2 == 0
filter_result = filter(is_even, [1, 2, 3, 4, 5, 6]) #Filtering even numbers from the list
print("Filter Result:", list(filter_result))
```
